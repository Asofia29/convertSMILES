```python
# Importar a biblioteca RDKit
!pip install rdkit-pypi
from rdkit import Chem

# Carregar a estrutura da molécula a partir de um arquivo (por exemplo, .mol ou .sdf)
# Exemplo usando um arquivo .mol
mol = Chem.MolFromMolFile('AK81.mol')

# Verificar se a molécula foi carregada corretamente
if mol:
    # Converter a estrutura em SMILES
    smiles = Chem.MolToSmiles(mol)
    print(f"SMILES: {smiles}")
else:
    print("Erro: A estrutura da molécula não pôde ser carregada.")
```

```python
# a partir de um arquivo SDF
suppl = Chem.SDMolSupplier('caminho_para_o_arquivo/exemplo.sdf')
for mol in suppl:
    if mol:
        smiles = Chem.MolToSmiles(mol)
        print(f"SMILES: {smiles}")
```

```python
# a partir de um arquivo PDB
from rdkit.Chem import AllChem

mol = Chem.MolFromPDBFile('caminho_para_o_arquivo/exemplo.pdb')
if mol:
    AllChem.Compute2DCoords(mol)  # Gera coordenadas 2D se necessário
    smiles = Chem.MolToSmiles(mol)
    print(f"SMILES: {smiles}")
```
